```python
def add(*num):    #single star for non-keyword arguments
    sum=0
    for n in num:
        sum +=n
    return sum
print(add())    # 0argument
print(add(1))   # 1argument
print(add(1,2))  # 2argument
print(add(1,2,3)) # 3argument
print(add(1,2,3,4)) # 4argument
```
